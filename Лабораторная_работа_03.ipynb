## Задание

Напишите функцию, которая по трем переданным им числам,
возвращает максимальное из них

## Решение

### Объявление функции

```python
def sum_ab(a, b):
  s = 0
  for i in range(a, b+1):
    s+=i*i
  return s
```

### Демонстрация использования функции

```python
x, y = 1, 2
print(f'Сумма квадратов числел от {x} до {y}: {sum_ab(x, y)}')
x, y = 500, 50000
print(f'Сумма квадратов числел от {x} до {y}: {sum_ab(x, y)}')
x, y = 268, 3258
print(f'Сумма квадратов числел от {x} до {y}: {sum_ab(x, y)}')
```

2. Библиотечная функция — это готовая функция, написанная заранее и хранящаяся в библиотеке (модуле), которую можно подключить и использовать в своей программе, не реализуя её заново. Примеры: math.sqrt(), random.randint(), len(), print().

3. Библиотечные функции похожи на «чёрные ящики», потому что программисту не нужно знать, как функция устроена внутри — достаточно знать, что она делает, какие аргументы принимает и что возвращает. Внутренняя реализация скрыта, как содержимое чёрного ящика: виден только «вход» и «выход».
