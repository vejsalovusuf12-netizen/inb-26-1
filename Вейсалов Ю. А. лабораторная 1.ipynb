## Задание

Необходимо вычислить значение выражения:
$$c=gh+b\cdot q3-x + y / w$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $gh, b, g3, x, y, w$. Переменная, которая должна быть вычислена на основе независимых: $c$.

```python
gh=10
b=30.67
q3=5
x=5
y=7
w=5
print("gh = ", gh, ", type(gh) = ", type(gh))
print("b = ", b, ", type(b) = ", type(b))
print("q3 = ", q3, ", type(q3) = ", type(q3))
print("x = ", x, ", type(x) = ", type(x))
print("w = ", w, ", type(w) = ", type(w))
print("y = ", y, ", type(y) = ", type(y))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $c$:

```python
c=gh + b * q3 - x + y / w
print('c=',c)
print('Type of c: ', type(c))
```
