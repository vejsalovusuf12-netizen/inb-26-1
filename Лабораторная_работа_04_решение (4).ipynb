## Задание

В файле data_v1.txt найдите количество чисел, кратных 3, но не кратных 9 (под числом понимается непрерывная последовательность цифр). В ответе укажите количество таких чисел и максимальное из них.

## Чтение файла

Читаю весь файл в строку `s`, лишние пробелы и переводы строк по краям убираю методом `strip()`.

```python
f=open('data_v1.txt')
s=f.read().strip()
print(len(s))
```

## Решение

Обхожу строку посимвольно и накапливаю цифры в `target`. Когда встречается буква, перевожу `target` в `int` и проверяю: число кратно 3 (`x % 3 == 0`) и не кратно 9 (`x % 9 != 0`). Подходящие числа считаю в `count` и ищу максимум в `maxx`. После цикла отдельно проверяю последнее число, если строка заканчивается цифрами.

```python
count=0
maxx=None
target=''
for c in s:
  if c.isdigit():
    target+=c
  else:
    if target!='':
      x = int(target)
      if x%3==0 and x%9!=0:
        count+=1
        if maxx is None or x>maxx:
          maxx=x
      target=''

if target!='':
  x = int(target)
  if x%3==0 and x%9!=0:
    count+=1
    if maxx is None or x>maxx:
      maxx=x

print('Количество чисел, кратных 3, но не кратных 9 = %d' % count)
print('Максимальное из них = %d' % maxx)
```

**Ответ:** количество чисел — `2766`, максимальное из них — `73758`.

## Ответы на контрольные вопросы

1. Функции `read()`, `readline()` и `readlines()` предназначены для чтения данных из текстового файла. Сходство: все три метода вызываются у файлового объекта, возвращают прочитанный текст в виде строк (или списка строк), сдвигают позицию считывания вперёд, а символ перевода строки `\n` в прочитанном тексте сохраняется. Различия: `read()` читает весь файл целиком (или указанное в скобках количество символов) и возвращает одну строку; `readline()` читает только одну строку файла, начиная с текущей позиции, и при каждом вызове переходит к следующей, а когда файл закончился, возвращает пустую строку; `readlines()` читает все оставшиеся строки файла и возвращает их списком, где каждый элемент — одна строка.

2. Многострочный файл можно прочитать, обойдя файловый объект циклом `for`: `for line in f:`. Файловый объект является итерируемым, поэтому на каждой итерации цикла переменная `line` получает очередную строку файла, и вызывать `read()`, `readline()` или `readlines()` явно не нужно. Обычно такой цикл записывают внутри блока `with open(...) as f:`.

3. Закрытие файла (метод `close()` или выход из блока `with`) освобождает файл и ресурсы операционной системы, которые были выделены для работы с ним. При записи закрытие также сбрасывает содержимое буфера на диск, поэтому записанные данные гарантированно сохраняются в файле. Кроме того, закрытый файл становится доступен другим программам, а количество одновременно открытых файлов не превышает допустимого предела.

4. Позиция считывания — это указатель (смещение в байтах или символах от начала файла), который показывает, с какого места файла будет прочитан следующий символ или строка. После каждой операции чтения она сдвигается на количество прочитанных данных. Когда файл открывается для чтения, позиция считывания находится в самом начале файла, то есть равна нулю. Вернуть её на нужное место можно методом `seek()`.

5. Чтобы записывать данные в файл, не удаляя его существующее содержимое, файл нужно открыть в режиме дозаписи — `'a'` (append). Новые данные при этом записываются в конец файла, после уже имеющегося текста. Для сравнения, режим `'w'` стирает прежнее содержимое файла.

6. Исключение — это ошибка, возникающая во время выполнения программы (например, деление на ноль, обращение к несуществующему файлу, неверное преобразование типов). В момент её возникновения нормальное выполнение программы прерывается, и интерпретатор ищет код, который сможет обработать эту ошибку. Исключения представлены объектами специальных классов (`ZeroDivisionError`, `FileNotFoundError`, `ValueError` и др.).

7. Если исключение возникло и в программе нет инструкции `try/except`, которая его обрабатывает, выполнение программы немедленно прекращается (программа аварийно завершается). На экран выводится сообщение об ошибке с трассировкой стека (traceback), в котором указаны тип исключения, текст ошибки и строки кода, где она произошла. Остальной код после места ошибки не выполняется.

8. При попытке открыть для чтения несуществующий файл программа вызывает исключение `FileNotFoundError` (в старых версиях Python — `IOError`).

9. При попытке преобразовать нечисловую строку в число функцией `float()` (например, `float('abc')`) программа вызывает исключение `ValueError`.

10. Если файл не существует и его открывают в режиме дозаписи `'a'`, ошибки не возникает: Python автоматически создаёт новый пустой файл с таким именем, и данные записываются в него.
